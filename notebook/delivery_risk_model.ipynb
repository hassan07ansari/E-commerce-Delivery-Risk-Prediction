```python
items_products = order_items.merge(products,on='product_id',how='left')
```

```python
print(items_products.shape)
print(items_products.head())
print(items_products.columns)
```

```python
items_products
```

```python
order_level_items = (items_products.groupby('order_id').agg(item_count=("order_item_id","count"),
total_price = ("price","sum"),avg_item_price = ("price","mean"),total_freight = ("freight_value","sum"),
total_weight_g = ("product_weight_g","sum"),avg_weight_g = ("product_weight_g","mean"),unique_products = ("product_id","nunique"),
unique_sellers = ("seller_id","nunique"),unique_categories = ("product_category_name","nunique")).reset_index())
```

```python
order_level_items
```

```python
final_df = orders_customers.merge(order_level_items,on="order_id",how="left")
```
